## Cluster geographic locations of deaths to identify hotspots using KMeans

#### 1. Extract latitude and longitude as numeric columns from DeathCityGeo.

```python
from sklearn.cluster import KMeans

# make a geo_data dataframe
geo_data = data[['DeathCityGeo']].dropna()

# look at how the death city geo data is structured

geo_data['DeathCityGeo'].head()

# split the data into latitude and longitude
geo_data['Latitude'] = geo_data['DeathCityGeo'].apply(lambda x: x.split(',')[0])
geo_data['Longitude'] = geo_data['DeathCityGeo'].apply(lambda x: x.split(',')[1])

```

```python
import pandas as pd
from sklearn.cluster import KMeans
import folium
import numpy as np

# Ensure DeathCityGeo data is available
geo_data = data[['DeathCityGeo']].dropna()

# Extract latitude and longitude
geo_data['Latitude'] = geo_data['DeathCityGeo'].apply(lambda x: float(x.split('(')[-1].split(',')[0]))
geo_data['Longitude'] = geo_data['DeathCityGeo'].apply(lambda x: float(x.split(',')[-1].split(')')[0]))

# Display the first few rows to confirm
print(geo_data.head())
```

#### 2. Cluster the Data Use KMeans to find clusters (hotspots) based on the latitude and longitude coordinates.

Each geographic location is assigned to a cluster based on its proximity to cluster centers.
kmeans.cluster_centers_ provides the central coordinates of each cluster (hotspots).
Interactive Map:

Points are plotted with folium, color-coded by cluster.
Cluster centers are marked to show the central hotspots.

```python
# Number of clusters (hotspots) - you can adjust this
n_clusters = 9

# Run KMeans on latitude and longitude
kmeans = KMeans(n_clusters=n_clusters, random_state=30)
geo_data['Cluster'] = kmeans.fit_predict(geo_data[['Latitude', 'Longitude']])

# Add cluster centers to analyze
centers = kmeans.cluster_centers_
print("Cluster Centers (Lat, Lon):", centers)
```

#### 3. Visualization on a map

```python
# Initialize a Folium map centered at the mean of the latitudes and longitudes
map_center = [geo_data['Latitude'].mean(), geo_data['Longitude'].mean()]
map_clusters = folium.Map(location=map_center, zoom_start=8)

# Add points for each location and color them based on their cluster
colors = ['red', 'blue', 'green', 'purple', 'orange', 'darkred', 'pink', 'yellow']
for i, row in geo_data.iterrows():
    folium.CircleMarker(
        location=(row['Latitude'], row['Longitude']),
        radius=5,
        color=colors[row['Cluster'] % len(colors)],
        fill=True,
        fill_opacity=0.5
    ).add_to(map_clusters)

# Add cluster centers as larger markers
for center in centers:
    folium.Marker(
        location=center,
        popup=f"Cluster Center: {center}",
        icon=folium.Icon(color="black", icon="info-sign")
    ).add_to(map_clusters)

# Save or display the map
map_clusters.save('geo_clusters_map.html')
map_clusters
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import silhouette_score

# Determine the optimal number of clusters using the Elbow Method
inertia = []
silhouette = []
range_n_clusters = range(2, 10)  # Test clusters from 2 to 9

for k in range_n_clusters:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(geo_data[['Latitude', 'Longitude']])
    inertia.append(kmeans.inertia_)
    silhouette.append(silhouette_score(geo_data[['Latitude', 'Longitude']], kmeans.labels_))

# Plot the Elbow Curve
plt.figure(figsize=(10, 5))
plt.plot(range_n_clusters, inertia, marker='o')
plt.title('Elbow Method for Optimal Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.show()

# Plot Silhouette Scores
plt.figure(figsize=(10, 5))
plt.plot(range_n_clusters, silhouette, marker='o', color='orange')
plt.title('Silhouette Scores for Optimal Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.show()
```
